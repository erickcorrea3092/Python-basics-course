```python
# prompt: genera un clasificador donde calsifique nadadores dependiendo de cuantos kilometros nada por semana, has tres clases principiante, intermedio y avanzado

def clasificar_nadador(km_semana):
  """
  Clasifica a un nadador según los kilómetros que nada por semana.

  Args:
    km_semana: Número de kilómetros nadados por semana.

  Returns:
    La clasificación del nadador ('Principiante', 'Intermedio', 'Avanzado').
  """
  if km_semana < 5:
    return "Principiante"
  elif km_semana < 15:
    return "Intermedio"
  else:
    return "Avanzado"

# Ejemplo de uso:
kilometros_nadados = float(input('Ingresa los kilómetros nadados por semana: '))
clasificacion = clasificar_nadador(kilometros_nadados)
print(f'La clasificación del nadador es: {clasificacion}')
```

```python
# prompt: genera un clasificador donde calsifique nadadores dependiendo de cuantos kilometros nada por semana, has tres clases principiante, intermedio y avanzado

def clasificar_nadador(km_por_sem):
  """
  Clasifica a un nadador según la cantidad de kilómetros que nada por semana.

  Args:
    km_por_sem: Número de kilómetros nadados por semana.

  Returns:
    Una cadena que indica la clasificación del nadador: 'Principiante',
    'Intermedio', 'Avanzado' o 'Excelente'.
  """
  if km_por_sem < 10:
    return 'Principiante'
  elif km_por_sem < 25:
    return 'Intermedio'
  elif km_por_sem < 50:
    return 'Avanzado'
  else:
    return 'Excelente'

# Ejemplo de uso:
km_nadados = 18
clasificacion = clasificar_nadador(km_nadados)
print(f"Con {km_nadados} km por semana, eres un nadador: {clasificacion}")

km_nadados = 5
clasificacion = clasificar_nadador(km_nadados)
print(f"Con {km_nadados} km por semana, eres un nadador: {clasificacion}")

km_nadados = 40
clasificacion = clasificar_nadador(km_nadados)
print(f"Con {km_nadados} km por semana, eres un nadador: {clasificacion}")
```

```python
import math
print('Programa de cálculo de raíz cuadrada')
numero=int(input('Introduce un número por favor: '))

intentos=0

while numero<0:
  print('No se puede hallar la raíz de un número negativo')
  if intentos==2: # = operador de asignación y el == se interpreta como 'igual a'
    print('Has consumido demasiados intentos. El programa ha finalizado')
    break;
  numero=int(input('Introduce un número por favor: '))
  if numero<0:
    #intentos=intentos+1
    intentos += 1

if intentos<2:
  solucion=math.sqrt(numero)
  print('La raíz cuadradada de ' +str(numero)+' es '+str(solucion))
```
