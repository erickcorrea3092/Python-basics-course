```python
def sum_resta(dato_1, dato_2):
  """se calcula la suma y resta de dos datos dados"""
  suma = dato_1 + dato_2
  resta = dato_1 - dato_2
  producto = dato_1 * dato_2
  return suma, resta, producto
```

```python
r = sum_resta(4,5)
print(r)
```

```python


r1= sum_resta(3,5)
r2= sum_resta(7,6)
r3= sum_resta(5,6)

print(r1)
print(r2)
print(r3)
```

Reto 10 Reto 10: En 10 minutos, crea una función que:

Reciba dos numeros.
Realice la multiplicación de ambos números.
Realice la exponenciación del número más grande como base y el más pequeño como exponente.
Regrese los valores de las operaciones anteriores y los imprima en pantalla.

```python
def f (a,b):
  if a>b:
    mayor=a
    menor=b
  else:
    mayor=b
    menor=a

  m = mayor*menor
  e = mayor**menor

  return m,e
```

```python
a = int(input('Dame el primer número: '))
b = int(input('Dame el segundo número: '))

m1 = f(a,b)
print('El resultado es ' + str(m1))
```
