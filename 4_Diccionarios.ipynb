Los diccionarios son agrupaciones de datos que mantienen la siguiente relación:

variable = {'idetificador1': 'valor1', 'idetificador2': 'valor2', 'idetificador3': 'valor3'}

de manera general

dict = {'keys': 'values'}

En este caso tenemos el diccionario nombrado como variable, sabemos que es un diccionario por el uso de los signos {:}. En este caso el diccionario variable tiene tres elementos, donde cada elemento cuenta con un identificador y un valor, los cuales pueden ser invocados y (re)utilizados en cualquier parte del código.

Vamos a utilizar diccionarios, haremos algunas operaciones sencillas con funciones y cadenas de texto

```python
def suma_poblaciones(Entrada):
  suma=0
  población = {'Nuevo León': 5.12, 'Chiapas': 5.218,
               'Guanajuato': 5.854, 'Puebla':6.169,
               'Veracruz': 8.113, 'Jalisco':8.256,
               'Estado de México': 16.19}
##población = {estado:pob}
#______CREAMOS EL BUCLE________________________
  for estado, pob in población.items(): #bucle for
    if Entrada[0] == estado or Entrada[1] == estado: #condicional if
      suma += pob
  return suma
```

```python
Entrada = []
Entrada.append(input('Dame el primer estado: '))
Entrada.append(input('Dame el segundo estado: '))
print('La suma es: ' + str(suma_poblaciones(Entrada))+
      ' milones') #genera el string (cadena)
```

```python
#otro ejemplo de función de suma de los valores para distintas claves

def fiesta(chavales):
  s=0
  I={'Juan':50, 'Mari': 70, 'Yaque':80, 'Adrian': 40, 'Lola': 30, 'Pedro':20}
  for c,v in I.items():
    if chavales[0] == c or chavales[1] == c:
      s += v
  return s

chavales = []
chavales.append(input('¿Quien va a ir?: '))
chavales.append(input('¿Alguien más?: '))
print('Tenemos '+str(fiesta(chavales))+' pesos para el pastel')
print(fiesta(chavales))
```

En Python, %s y %d son especificadores de formato (forma) que se utilizan para formatear (dar formato) cadenas y números, respectivamente:

%s: Es un operador que se usa para formatear cadenas con texto.

%d: Es un especificador de argumento que se utiliza para representar valores numéricos.


```python
peliculas = {'Hellboy':'Guillermo del Toro', 'Roma':'Alfonso Cuarón'}
codigos = {'Mexico':52, 'Brasil': 55,'Francia': 33, 'Panamá':507 }
```

```python
for pais,cod in codigos.items(): #usamos un bucle que nos ayudará a generar strings
  print('%s tiene el código: %d' % (pais,cod)) #genera el string (cadena)
```

```python
for pelicula,director in peliculas.items():
  print('La película %s fue drigida por %s' % (pelicula,director))
```

```python
#comparación entre diccionarios

c1 = {'Brasil': 'Brasilia','Francia':'París', 'México':'CDMX'}
c2 = {'Brasil':'Brasilia','EUA':'Washington','México':'CDMX'}
c3 = {'México':'CDMX','Brasil':'Brasilia','Francia':'París'}

print(c1==c2) #False
print(c1==c3) # ? no importa el orden
print(c2!=c3) #True #c2 is not equal to c3
print(c2==c3) #False
```

#modificación de diccionarios

```python
alumnos = {'Jesús':10, 'Manuel':7,'Raúl':8,'Tania':9} #definimos el diccionario
print('El diccionario original es: ')
print(alumnos)
#________________________________________
print('___________')
print('__modificamos la calificación de Jesús, hizo trampa XC___')
alumnos['Jesús'] = 5
print(alumnos)
```

```python
#______________________________________________
print('___________')
print('__Añadimos la calificación de Alejandra___')
alumnos['Alejandra'] = 6
print (alumnos)
```

```python
#_______________________________________________
print('___________')
print('__Eliminamos a Manuel, no está inscrito :(___')
del alumnos['Manuel']
print(alumnos)
```
