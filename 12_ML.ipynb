```python
# Cargando bibliotecas
from sklearn.datasets import load_digits
from sklearn.manifold import TSNE
import matplotlib.pyplot as plt
from sklearn.decomposition import PCA
import plotly.express as px
import seaborn as sns
```

```python
# Cargando el dataset
digits = load_digits()

```

#Reducción de dimensionalidad por PCA

```python
# Creando un estimador para la reducción de dimensionalidad con PCA
pca = PCA(n_components=2)

# Transformando las características del dataset
reduced_data_pca = pca.fit_transform(digits.data)
print(f"Dimensiones originales: {digits.data.shape}")
print(f"Dimensiones reducidas con PCA: {reduced_data_pca.shape}")
```

```python
# Visualizando los datos reducidos con diferentes colores para cada digito
dots = plt.scatter(reduced_data_pca[:, 0], reduced_data_pca[:, 1],
     c=digits.target, cmap=plt.cm.get_cmap('nipy_spectral_r', 10))

colorbar = plt.colorbar(dots)
```

```python
# Visualización PCA con Plotly
fig_pca_plotly = px.scatter(x=reduced_data_pca[:, 0], y=reduced_data_pca[:, 1],
                             color=digits.target.astype(str),
                             color_continuous_scale=px.colors.sequential.Viridis,
                                             height=500, width=500,
                             title='Visualización PCA de dígitos manuscritos',
                             labels={'x': 'Componente 1 PCA', 'y': 'Componente 2 PCA', 'color': 'Dígito'})
fig_pca_plotly.show()
```

#Reducción de dimensionalidad por TSNE

```python
# Creando un estimador para la reducción de dimensionalidad
tsne = TSNE(n_components=2)
# Transformando las caracteristicas del dataset a dos dimensiones
reduced_data = tsne.fit_transform(digits.data)
digits.data.shape, reduced_data.shape
```

```python

# Visualizando los datos reducidos con diferentes colores para cada digito
dots = plt.scatter(reduced_data[:, 0], reduced_data[:, 1],
     c=digits.target, cmap=plt.cm.get_cmap('nipy_spectral_r', 10))

colorbar = plt.colorbar(dots)
```

```python
fig = px.scatter(x=reduced_data[:, 0], y=reduced_data[:, 1],
                 color=digits.target.astype(str),
                 color_continuous_scale=px.colors.sequential.Viridis,
                                             height=600, width=500,
                 title='Visualización t-SNE de dígitos manuscritos',
                 labels={'x': 'Componente 1 t-SNE', 'y': 'Componente 2 t-SNE', 'color': 'Dígito'})
fig.show()
```

#Reducción de dimensionalidad por UMAP

```python
import umap

# Creando un estimador para la reducción de dimensionalidad con UMAP
umap_reducer = umap.UMAP(n_neighbors=10, n_components=2, random_state=42)

# Transformando las características del dataset
reduced_data_umap = umap_reducer.fit_transform(digits.data)

print(f"Dimensiones originales: {digits.data.shape}")
print(f"Dimensiones reducidas con UMAP: {reduced_data_umap.shape}")
```

```python
# Visualizando los datos reducidos con diferentes colores para cada digito
dots = plt.scatter(reduced_data_umap[:, 0], reduced_data_umap[:, 1],
     c=digits.target, cmap=plt.cm.get_cmap('nipy_spectral_r', 10))

colorbar = plt.colorbar(dots)
```

```python
# Visualización UMAP con Plotly
fig_umap_plotly = px.scatter(x=reduced_data_umap[:, 0], y=reduced_data_umap[:, 1],
                             color=digits.target.astype(str),
                             color_continuous_scale=px.colors.sequential.Viridis,
                             height=500, width=500,
                             title='Visualización UMAP de dígitos manuscritos',
                             labels={'x': 'Componente 1 UMAP', 'y': 'Componente 2 UMAP', 'color': 'Dígito'})
fig_umap_plotly.show()
```
