```python
crp = np.array([5.722, 3.992, 2.755, 2.338, 2.137, 1.984, 1.747, 1.647, 1.457, 1.438])

print('La suma es: %f' %crp.sum())
print('El mínimo es: %f' %crp.min())
print('El máximo es: %f' %crp.max())
print('La media es: %f' %crp.mean())
print('La desviación estándar es: %f' %crp.std())
print('La varianza es: %f' %crp.var())
```

#Operaciones de arreglos con bucles

```python
import numpy as np

a = ([])
b = ([])
#usamos range para una crear serie de datos y creamos el bucle for in, para realizar la operacón a cada uno de lo daos
for i in range(2): #genera filas
  for j in range (2): #genera columnas
    v = float(input('escribe los datos: ')) #Ingresa los datos (input), lo tranforma a decimal y lo guarda en v
    if i == 1:  #esta condición funcionará solo para los indices 0
      a = np.append(a,v) # la fuunción np.append(a,v) solo servira para cuando el índice vale 1, primero deben llenarse los indices 0
    else:
      b = np.append(b,v) #con la función np.append(b,v) se llenarán los datos con el index 0 y otros distintos a 1


print(a)
print(b)
```

```python
print(a==b)
print(a!=b)
print(a>b)
print(a<b)
print(a>=b)
print(a<=b)
```
