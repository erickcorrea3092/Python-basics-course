# Datos desde arreglos (arrays)

Si matplotlib se limitara a trabajar con listas, sería bastante inútil para el procesamiento numérico. Generalmente, se utilizan matrices numpy . De hecho, todas las secuencias se convierten en matrices numpy internamente. El siguiente ejemplo ilustra cómo representar tres diferentes trazas, con líneas de diferentes estilos de formato.

```python
# Generamos los t datos, en forma de array de numpy en un rango de 0 a 5 con pasos de 0.2,
# es decir 0.0, 0.2, 0.4, 0.6,0.8.......
t = np.arange(0., 5., 0.2)

# Definimos tres trazas y serán representadas respectivamente mediante:
# lineas sesgadas rojas, cuadrados(square) azules y triángulos magneta
plt.plot(t, t, 'r--', t, t**2-12, 'bs', t, 2*t**3+5, 'm^') # 3 trazas en una función plt
plt.show()
```

```python
# Vamos a representar funciones trigonométricas

# Generamos datos triginométricos
v1 = -2*np.pi #-2.pi
v2 = 2*np.pi #2.pi
i = 100

# Crear arreglo con función linspace: np.linspace(start, stop, num)
t = np.linspace(float(v1),float(v2), int(i))

# Generamos las gráficas, tres trazas con tres funciones
plt.plot(t,np.cos(t),'bo') #traza 1
plt.plot(t,np.sin(t),'-r') #traza 2
plt.plot(t,np.tan(t),'-g') #traza 3
plt.plot()
```

```python
#Podemos realizar un histograma para graficar frecuecnia
# Datos de ejemplo
datos = np.random.randn(1000)  # Genera 1000 números aleatorios con distribución normal

# Crear el histograma
plt.hist(datos, bins=81, color = 'y')  # 'bins' define el número de barras en el histograma

# Agregar etiquetas y título
plt.xlabel("Valores")
plt.ylabel("Frecuencia")
plt.title("Histograma de datos")

# Mostrar el gráfico
plt.show()
```

# Gráficos desde Dataframes

Es bastante más común generar gráficas desde datos estructurados como diccionarios o matrices de pandas conocidas como DataFrames. En el siguiente ejemplo vamos a generar datos sinteticos desde funciones de numpy (variables 'a', 'c' y 'd'), que representaría las características de nuestros datos y los vamos a almacenar en el diccionario data. Vamos a generar una nueva serie de datos 'b' que se van a generar, en función de la variable 'a' y vamos a realizar una transformación a los datos de la variable 'd', obteniendo el valor absoluto con la función np.abs y multiplicando por cien (*100)

```python
# Generamos diccionario con datos aleatorios
data = {'a': np.arange(50),# genera un array de números enteros ordenados desde el 0 al 50
        'c': np.random.randint(0, 50, 50), # genera un array de 50 números enteros aleatorios desde el 0 al 50
        'd': np.random.randn(50)} #genera un array de 50 números decimales aleatorios
                                #distribuidos según una distribución normal estándar (media 0, desviación estándar 1).

# Generamos 'b' en función de 'a' y la transformación de 'd'
data['b'] = data['a'] + 10 * np.random.randn(50)
data['d'] = np.abs(data['d']) * 100
# Nos podemos cerciorar de que data es un dicionario
#data
```

```python
#Muchas veces es más fácil visualizar la información en forma de DataFrames
data_df = pd.DataFrame(data)
print(data_df.shape)
data_df#.head()
```

Hemos generado datos aleatorios que hemos organizado en un diccionario y también en un Dataframe. Esto nos representaría que nosotros tenemos 50 elementos (que pudieran ser personas, niños, casas, moléculas, pacientes, étc.) con 4 características, en donde cada característica nos representa una dimensión. Matplotlib nos permite representar estos elementos en un plano cartesiano, los ejes estarían representando a estas dimensiones o características, en este ejemplo vamos a usar las variables 'a' y 'b' para hacer la representación de dimensiones con los ejes. Podemos utilizar un tercer eje para representar una tercer característica en un espacio tridimensional o podemos utilizar otras propiedades de las gráficas como el color y el tamaño de los markers o puntos. En este ejemplo usaremos el color para representar la caracteristica 'c', que sería nuestra tercera dimensión y la característica 'd', que sería nuestra cuarta dimensión con el tamaño (size)

```python
plt.scatter('a', #variable x
            'b', #variable y
            c='c', #color
            s='d', #size
            data=data #asigna los datos puede ser desde el diccionario o desde el dataframe
            )
plt.xlabel('entry a')
plt.ylabel('entry b')
plt.show()
```

```python
names = ['group_a', 'group_b', 'group_c']
values = [1, 10, 100]


plt.figure(figsize=(9, 3)) #parametros del tamño de figura

plt.subplot(131)
plt.bar(names, values)
# plt.grid(True)
#plt.plot(names, values)
#_________________________________________________________
plt.subplot(132)
plt.scatter(names, values)
plt.grid(True)
#_________________________________________________________
plt.subplot(133)
plt.plot(names, values, 'r*-')

#_________________________________________________________
plt.suptitle('Categorical Plotting')
plt.show() #algoritmo para mostrar
```

```python
plt.figure(figsize=(4, 9)) #parametros del tamño de figura

plt.subplot(311)
plt.bar(names, values)
plt.grid(True)
#plt.plot(names, values)
#_________________________________________________________
plt.subplot(312)
plt.scatter(names, values)
plt.grid(True)
#_________________________________________________________
plt.subplot(313)
plt.plot(names, values, 'r*-')
#_________________________________________________________
plt.suptitle('Categorical Plotting')
plt.show() #algoritmo para mostrar
```

```python
# plt.subplot(131) es una función de Matplotlib que se utiliza para crear un subgráfico dentro de una figura. Analicemos qué significa cada parte:

# **plt.subplot()**: Esta es la función principal. Se utiliza para dividir una sola figura en una cuadrícula de filas y columnas y, luego, seleccionar una
#subgráfica específica dentro de esa cuadrícula con la que trabajar.

# **`131`**: Esta es una notación abreviada que representa la configuración de la cuadrícula y la subgráfica específica que estás eligiendo.
#Se interpreta de la siguiente manera:
    # **`1`**: El número de filas en la cuadrícula (1 fila en este caso).
    # **`3`**: El número de columnas en la cuadrícula (3 columnas).
    # **`1`**: El índice de la subgráfica que desea activar (la primera subgráfica en la cuadrícula 1x3).

#**En esencia:** `plt.subplot(131)` crea una figura con una sola fila y tres columnas de subgráficas, y luego selecciona la primera de esas tres subgráficas
#como el área de representación gráfica activa. Cualquier comando de representación gráfica posterior (como `plt.bar`, `plt.plot`, etc.) afectará
#*solo* esa primera subgráfica.

#**Ejemplo en su código:** Su código está creando una figura con tres subgráficas dispuestas horizontalmente.

# 1. `plt.subplot(131)`: selecciona la primera subgráfica. Luego, `plt.bar(names, values)` crea un gráfico de barras en esta subsección.
# 2. `plt.subplot(132)`: *Esta línea seleccionaría la segunda subsección. Luego, `plt.scatter(names,values)` crearía un gráfico de dispersión en esta área.
# 3. `plt.subplot(133)`: *Esta línea selecciona la tercera subgráfica. `plt.plot(names,values)` crea una gráfica lineal en la tercera subgráfica.

#**Para ver las tres gráficas, descomente esas líneas.**
# La figura resultante mostrará un gráfico de barras, un gráfico de dispersión y un gráfico lineal, cada uno en su propia subgráfica dentro de la misma ventana
# de la figura.
```

```python
def f(t):
    return np.exp(-t) * np.cos(2*np.pi*t)

t1 = np.arange(0.0, 5.0, 0.1)
t2 = np.arange(0.0, 5.0, 0.02)

plt.figure()
plt.subplot(211) #dos filas, una columna, indice 1
plt.plot(t1, f(t1), 'go', t2, f(t2)-3, 'k') #aquí definimos dos trazas

plt.subplot(212) #dos filas, una columna, indice 2
plt.plot(t2, np.cos(2*np.pi*t2), 'r--') # definimos una traza
plt.show()
```

Si quieres aprender más a usar esta función subplot te dejo el siguiente ejemplopara que pruebes lo que se puede hacer cambiando el tamaño de columnas, filas e indices, así como el [manual](https://matplotlib.org/stable/users/explain/axes/tight_layout_guide.html#sphx-glr-users-explain-axes-tight-layout-guide-py)

```python
def f(t):
    return np.exp(-t) * np.cos(2*np.pi*t)


t1 = np.arange(0.0, 3.0, 0.01)

ax1 = plt.subplot(212)
ax1.margins(0.05)           # Default margin is 0.05, value 0 means fit
ax1.plot(t1, f(t1))

ax2 = plt.subplot(221)
ax2.margins(2, 2)           # Values >0.0 zoom out
ax2.plot(t1, f(t1))
ax2.set_title('Zoomed out')

ax3 = plt.subplot(222)
ax3.margins(x=-0, y=-0.25)   # Values in (-0.5, 0.0) zooms in to center
ax3.plot(t1, f(t1))
ax3.set_title('Zoomed in')

plt.show()
```

# Trabajando con texto

[text](https://matplotlib.org/stable/api/_as_gen/matplotlib.pyplot.text.html#matplotlib.pyplot.text) se puede utilizar para agregar texto en una ubicación arbitraria, y [xlabel](https://matplotlib.org/stable/api/_as_gen/matplotlib.pyplot.xlabel.html#matplotlib.pyplot.xlabel), [ylabel](https://matplotlib.org/stable/api/_as_gen/matplotlib.pyplot.ylabel.html#matplotlib.pyplot.ylabel) y [title](https://matplotlib.org/stable/api/_as_gen/matplotlib.pyplot.title.html#matplotlib.pyplot.title) se utilizan para agregar texto en las ubicaciones indicadas (consulte [Text in Matplotlib](https://matplotlib.org/stable/users/explain/text/text_intro.html#text-intro) para obtener un ejemplo más detallado)

```python
mu, sigma = 100, 15
x = mu + sigma * np.random.randn(10000)

# the histogram of the data
n, bins, patches = plt.hist(x, 500, density=True, facecolor='b', alpha=0.7)


plt.xlabel('Smarts')
plt.ylabel('Probability')
plt.title('Histogram of IQ')
plt.text(120, .02, r'$\mu=100,\ \sigma=15$') # agrega texto a un gráfico de matplotlib
plt.axis([40, 160, 0, 0.03]) #defino los limtes de los ejes x y y [x1,x2,y1,y2]
plt.grid(True) #True
plt.show()
```

```python
# La línea plt.text(60, .025, r'$\mu=100,\ \sigma=15$') agrega texto a un gráfico de matplotlib. Vamos a desglosarlo:

# **plt.text(...)**: Esta es la llamada de función para agregar texto al gráfico.
# **`60, .025`**: Estas son las coordenadas x e y donde se colocará el texto en el gráfico. Las unidades dependen de los límites de los ejes actuales.

# **`r'$\mu=100,\ \sigma=15$'`**: Esta es la cadena que se mostrará. El prefijo 'r' la convierte en una cadena sin formato, lo cual es importante para manejar
# las barras invertidas correctamente.
#Los `$...$` indican que la cadena contiene notación matemática LaTeX.

#* `$\mu$`: Esto muestra la letra griega mu (µ), que se usa a menudo para representar la media o el promedio en las estadísticas.
#* `$=100$`: muestra un signo igual y el valor 100.
#* `$,\ \sigma$`: muestra una coma, un espacio y la letra griega sigma (σ), que normalmente representa la desviación estándar en estadística.
#* `$=15$`: muestra un signo igual y el valor 15.

# En resumen, esta línea de código agrega el texto "µ=100, σ=15" al gráfico en las coordenadas especificadas (60, 0.025), utilizando el formato LaTeX para las letras griegas mu y sigma. Esto se utiliza normalmente para anotar un gráfico, por ejemplo, para mostrar la media y la desviación estándar de los datos que se muestran en el gráfico.
```

```python
# La línea `plt.axis([40, 160, 0, 0.03])` establece los límites de los ejes x e y del gráfico actual.

# **`[40, 160, 0, 0.03]`**: Esta lista define los límites de los ejes.
# `40`: El valor mínimo en el eje x.
# `160`: El valor máximo en el eje x.
# `0`: El valor mínimo en el eje y.
# `0.03`: El valor máximo en el eje y.

#En el código dado, limita la visualización del histograma al rango del eje x de 40 a 160 y al rango del eje y de 0 a 0.03.
#Esto se hace para lograr claridad visual y centrarse en una región específica de la distribución de datos, en lugar de mostrar todo el rango posible.
```

```python
# Vamos a colocar una felcha dentro de una gráfica

# Definimos los datos
t = np.arange(0.0, 5.0, 0.01) #variable independiente
s = np.cos(2*np.pi*t) #varable dependiente
plt.plot(t, s, lw=2) #lw = line width/ ancho de linea

#Realizamos las anotaciones
plt.annotate('local max', #legenda
             xy=(0, 1.0), #coordenadas punta de flecha
             xytext=(4.0, 1.5), #coordenadas cola de flecha
             arrowprops=dict(facecolor='black', shrink=0.05), #color y sseparación del 5% (0.05) con respecto al texto
             )

#Creaos la gráfica
plt.ylim(-2, 2) #define límites
plt.show()
```

# Transformaciones

También podemos hacer escalamiento de nuestros datos, múy útil en muchas disciplinas científicas

```python
# Fijamos un estado aleatorio (random state) para reprodubilidad en datos aleatorios
np.random.seed(19680801)

# make up some data in the open interval (0, 1)
y = np.random.normal(loc=0.5, scale=0.4, size=1000)
y = y[(y > 0) & (y < 1)]
y.sort()
x = np.arange(len(y))

# plot with various axes scales
plt.figure()

# linear
plt.subplot(221)
plt.plot(x, y)
plt.yscale('linear')
plt.title('linear')
plt.grid(True)

# log
plt.subplot(222)
plt.plot(x, y)
plt.yscale('log')
plt.title('log')
plt.grid(True)

# symmetric log
plt.subplot(223)
plt.plot(x, y - y.mean())
plt.yscale('symlog', linthresh=0.01)
plt.title('symlog')
plt.grid(True)

# logit
plt.subplot(224)
plt.plot(x, y)
plt.yscale('logit')
plt.title('logit')
plt.grid(True)
# Adjust the subplot layout, because the logit one may take more space
# than usual, due to y-tick labels like "1 - 10^{-3}"
plt.subplots_adjust(top=0.92, bottom=0.08, left=0.10, right=0.95, hspace=0.25,
                    wspace=0.35)

plt.show()
```

# Plotly

```python
import plotly.graph_objects as go
import numpy as np

# Ejercicio 1: Gráfico de dispersión simple
x = np.random.rand(100)
y = np.random.rand(100)
fig = go.Figure(data=go.Scatter(x=x, y=y, mode='markers'))
fig.update_layout(title='Gráfico de Dispersión', xaxis_title='Eje X', yaxis_title='Eje Y')
fig.show()


# Ejercicio 2: Gráfico de líneas con múltiples trazas
x = np.linspace(0, 10, 100)
y1 = np.sin(x)
y2 = np.cos(x)
fig = go.Figure()
fig.add_trace(go.Scatter(x=x, y=y1, mode='lines', name='Seno'))
fig.add_trace(go.Scatter(x=x, y=y2, mode='lines', name='Coseno'))
fig.update_layout(title='Gráfico de Líneas', xaxis_title='Tiempo', yaxis_title='Amplitud')
fig.show()


# Ejercicio 3: Histogramas
data = np.random.randn(1000)
fig = go.Figure(data=[go.Histogram(x=data)])
fig.update_layout(title='Histograma', xaxis_title='Valores', yaxis_title='Frecuencia')
fig.show()


# Ejercicio 4: Gráfico de barras
categories = ['A', 'B', 'C', 'D']
values = [25, 15, 30, 20]
fig = go.Figure(data=[go.Bar(x=categories, y=values)])
fig.update_layout(title='Gráfico de Barras', xaxis_title='Categorías', yaxis_title='Valores')
fig.show()


# Ejercicio 5: Gráfico de burbujas
x = np.random.rand(50)
y = np.random.rand(50)
size = np.random.rand(50) * 50
color = np.random.rand(50)
fig = go.Figure(data=go.Scatter(x=x, y=y, mode='markers', marker=dict(size=size, color=color, colorscale='Viridis')))
fig.update_layout(title='Gráfico de Burbujas', xaxis_title='Eje X', yaxis_title='Eje Y')
fig.show()
```
