# Clases


Una clase en Python es una plantilla o plano para crear objetos, que son instancias de esa clase. Las clases agrupan datos (atributos) y funciones (métodos) para definir un nuevo tipo de dato personalizado.

Características principales

Plantilla para objetos: Actúa como un molde que define cómo se crearán los objetos.
Atributos y métodos: Las clases contienen atributos (variables que almacenan el estado de un objeto) y métodos (funciones que definen el comportamiento del objeto).


Instancias: Un objeto es una instancia específica de una clase, como un mi_coche que es una instancia de la clase Coche. Cada instancia puede tener sus propios valores para los atributos (por ejemplo, un coche rojo y un coche azul).



Reutilización de código: Permiten agrupar datos y funcionalidades relacionados, haciendo que el código sea más organizado y fácil de reutilizar.
Programación Orientada a Objetos (POO): Las clases son un concepto fundamental de la POO, que también incluye principios como la herencia.

Python es un [lenguaje orientado a objetos](https://en.wikipedia.org/wiki/Object-oriented_programming). Un objeto es un espacio en la memoria que cuenta con **comportamiento**, **estado** e **identidad**.

* Comportamiento: Representado por métodos o funciones tales como `dormir()`, `correr()`, `saludar()` etc.
* Estado: Representado por atributos tales como `edad`, `nombre`, etc.
* Identidad: Al crear los objetos obtenemos una referencia del objeto para poder acceder a su estado y comportamiento.

El constructor __init__(): Es un método especial que se ejecuta automáticamente cada vez que se crea una nueva instancia de la clase.
Se utiliza para inicializar los atributos de un objeto.
El parámetro self siempre es el primer argumento, y hace referencia a la instancia de la clase que se está creando.

# Decoradores

En los lenguajes orientados a objetos es común encontrarnos con los conceptos de* getters* y *settlers*. Un getter es un método que permite acceder a los atributos de una clase. Un setter es un método que permite establecer el valor de los atributos en una clase.

Una forma de acceder y modificar es utilizando los decoradores** @property** y **@nombre_método.settter** el objetivo principal de cualquier decorador es modificar los métodos o atributos de su clase de tal manera que el usuario de su clase no necesite hacer ningún cambio en su código.

```python
@property
def nombre(self):
  return self._nombre_de_atributo
@nombre.setter
def nombre(self, nombre):
  self._nombre_de_atributo = nombre
```

Los programadores de Python a menudo usan atributos "privados" para datos o métodos de
utilidad que son esenciales para el funcionamiento interno de una clase pero que no forman parte de la
interfaz pública de la clase. Python tiene una convención de nomenclatura para atributos "privados",
utilizando dos guiones bajos (__) en el nombre del atributo.

# Ejemplo 4 🌡

El siguiente ejemplo define una clase `Temperatura` con un atributo `celsius`. Usamos `@property` para definir cómo se obtiene `celsius` y `@celsius.setter` para controlar cómo se establece, incluyendo una validación para asegurar que la temperatura no sea menor a -273.15 (cero absoluto).

```python
class Temperatura:
    def __init__(self, celsius):
        self._celsius = 0  # Inicializamos un valor por defecto
        self.celsius = celsius # Usamos el setter para la validación inicial

    @property
    def celsius(self):
        """El atributo celsius."""
        return self._celsius

    @celsius.setter
    def celsius(self, valor):
        if valor < -273.15:  # Cero absoluto
            raise ValueError("La temperatura no puede ser menor a -273.15 grados Celsius.")
        self._celsius = valor

    @property
    def fahrenheit(self):
        """La temperatura en grados Fahrenheit."""
        return (self.celsius * 9/5) + 32

    @fahrenheit.setter
    def fahrenheit(self, valor):
        self.celsius = (valor - 32) * 5/9  # Convertir Fahrenheit a Celsius y usar el setter de Celsius
```

```python
# --- Ejemplos de uso ---

# Crear una instancia de Temperatura
temp = Temperatura(25)
print(f"Temperatura inicial: {temp.celsius}°C, {temp.fahrenheit}°F")
```

```python
# Modificar la temperatura usando el setter de celsius
temp.celsius = 30
print(f"Nueva temperatura: {temp.celsius}°C, {temp.fahrenheit}°F")
```

```python
# Modificar la temperatura usando el setter de fahrenheit
temp.fahrenheit = 68 # Esto establecerá celsius a 20
print(f"Temperatura modificada por Fahrenheit: {temp.celsius}°C, {temp.fahrenheit}°F")
```

```python
# Intentar establecer una temperatura inválida (activará la validación del setter)
try:
    temp.celsius = -300
except ValueError as e:
    print(e)
```

# Ejemplo 5 🔵

Aquí tienes un buen ejemplo de cómo usar los decoradores @property y @setter con una clase Circulo.. Se verifica que la propiedad area es de solo lectura (gracias a que solo tiene @property y no @area.setter), lo que significa que no se puede asignar un valor directamente a ella, generando un AttributeError si se intenta.

```python
import math

class Circulo:
    def __init__(self, radio):
        self._radio = 0  # Inicializamos con un valor por defecto
        self.radio = radio  # Usamos el setter para la validación inicial

    @property
    def radio(self):
        """El radio del círculo."""
        return self._radio

    @radio.setter
    def radio(self, valor):
        if not isinstance(valor, (int, float)) or valor < 0:
            raise ValueError("El radio debe ser un número positivo.")
        self._radio = valor

    @property
    def area(self):
        """El área del círculo (calculada)."""
        return math.pi * (self.radio ** 2)
```

```python
# --- Ejemplos de uso ---

# Crear una instancia de Circulo
c1 = Circulo(5)
print(f"Círculo 1: Radio = {c1.radio}, Área = {c1.area:.2f}")
```

```python
# Modificar el radio usando el setter
c1.radio = 7.5
print(f"Círculo 1 (modificado): Radio = {c1.radio}, Área = {c1.area:.2f}")
```

```python
# La propiedad 'area' es de solo lectura, no se puede asignar directamente
try:
    c1.area = 50 # Esto generará un AttributeError
except AttributeError as e:
    print(f"Error al intentar asignar al área: {e}")
```

```python
# Intentar establecer un radio inválido
try:
    c1.radio = -2
except ValueError as e:
    print(f"Error al establecer radio: {e}")

try:
    c1.radio = "invalido"
except ValueError as e:
    print(f"Error al establecer radio: {e}")
```

```python
class Persona:
  def __init__(self, nombre, edad, sexo):
    if type(nombre) !=str or nombre == '':
      raise ValueError('nombre inválido')
    if edad<0 and edad !=int:
      raise ValueError('edad inválida')
    if sexo not in ['hombre', 'mujer']:
      raise ValueError('sexo inválido')
    self._nombre = nombre
    self._edad = edad

  @property
  def nombre(self):
    return self._nombre
  @property
  def edad(self):
    return self._edad
```

```python
class Estudiante(Persona):
  def __init__(self, nombre, edad, sexo, curso):
    Persona.__init__(self, nombre, edad, sexo)
    self._curso = curso

  @property
  def curso(self):
    return self._curso
```

```python
alumno1 = Estudiante('Jose',21,'mujer', 'Python para IA')
print(alumno1.nombre)
print(alumno1.edad)
print(alumno1.curso)
```

```python
class Persona(object):
  # constructor. cuando se crea un nuevo objeto se llamara a esta funcion
  def __init__(self, nombre, edad):
    # self.nombre, self.edad y self.energia son atributos del objeto persona
    self.nombre = nombre
    self.edad = edad
    self.energia = 10

  # dormir es un método de los objetos persona. self es el objeto persona
  def dormir(self):
    self.energia = 10
    print('zzz')

  def correr(self):
    if self.energia >= 5:
      print('Corriendo...')
      self.energia -= 2
    else:
      print('Estoy muy cansado para correr :(')

  def saludar(self):
    print('Hola, me llamo {} y tengo {} años'.format(self.nombre, self.edad))
```

Además en los lenguajes orientados a objetos existe un término conocido como herencia [herencia](https://en.wikipedia.org/wiki/Inheritance_(object-oriented_programming)). La herencia sirve para reutilizar comportamiento de una clase existente (como `Persona`) y expandirlo para que pueda hacer cosas más específicas, dependiendo de lo que se quiera resolver.

Por ejemplo, podemos definir una clase `Trabajador` que haga todo lo que pueda hacer una persona pero aparte pueda trabajar.

```python
# clase trabajador. De esta manera sabemos que hereda de la clase Persona
class Trabajador(Persona):
  # sobreescribimos el constructor para agregar la propiedad lugar_de_trabajo
  def __init__(self, nombre, edad, lugar_de_trabajo):
    # llamamos al constructor de Persona, este inicializa nombre y edad
    super().__init__(nombre, edad)
    self.lugar_de_trabajo = lugar_de_trabajo

  def trabajar(self):
    # dado que Trabajador hereda de Persona, puede acceder al atributo energia
    if self.energia >= 5:
      print('Estoy trabajando en {}'.format(self.lugar_de_trabajo))
    else:
      print('Estoy muy cansado para trabajar :(')
```

Ya existía un método constructor en la clase padre, sin embargo en `Trabajador` estamos definiendo un nuevo constructor. Esto es conocido como [sobreescritura de métodos](https://en.wikipedia.org/wiki/Method_overriding). Esta sobreescritura nos permite definir los métodos nuevamente para cambiar el comportamiento de funciones existentes, inicializando el atributo `lugar_de_trabajo` en este caso. La llamada a la función [`super`](https://docs.python.org/3/library/functions.html#super) nos permite correr métodos de instancia en el contexto de la clase padre (`Persona`); entonces, al llamar al constructor desde `super` ejecutará el constructor definido en `Persona`. Es por esto que se inicializan los atributos `nombre`, `edad` y `energia`. De este modo, `Trabajador` puede hacer todo lo que puede hacer `Persona`  (`saludar`, `dormir` y `correr`) pero también puede trabajar, cosa única para él.

```python
pedro = Trabajador('pedro', 31, 'Walmart')
pedro.saludar()
pedro.correr()#10-2
pedro.trabajar() #8
pedro.dormir() #10
pedro.correr() #10-2
pedro.correr() #8-2
pedro.correr()#6-2
pedro.trabajar()
```

```python
import math

class Personaje:
    """
    Clase base para cualquier personaje en el juego (jugador, enemigo, NPC).
    """
    def __init__(self, nombre, salud_maxima, pos_x, pos_y, velocidad):
        """Inicializa los atributos del personaje."""
        self.nombre = nombre
        self.salud_maxima = salud_maxima
        self.salud_actual = salud_maxima
        self.pos_x = pos_x
        self.pos_y = pos_y
        self.velocidad = velocidad
        self.esta_vivo = True

    def mover(self, delta_x, delta_y):
        """Actualiza la posición del personaje en el mapa."""
        if self.esta_vivo:
            self.pos_x += delta_x * self.velocidad
            self.pos_y += delta_y * self.velocidad
            print(f"{self.nombre} se movió a ({self.pos_x}, {self.pos_y}).")

    def recibir_dano(self, cantidad_dano):
        """Reduce la salud del personaje y comprueba si ha muerto."""
        if self.esta_vivo:
            self.salud_actual -= cantidad_dano
            print(f"{self.nombre} recibió {cantidad_dano} de daño. Salud actual: {self.salud_actual}")
            if self.salud_actual <= 0:
                self.salud_actual = 0
                self.esta_vivo = False
                print(f"{self.nombre} ha sido derrotado/a.")

    def atacar(self, objetivo):
        """Define la acción de ataque a otro personaje."""
        if self.esta_vivo and objetivo.esta_vivo:
            dano_infligido = 10 # Valor de daño base, se puede personalizar
            print(f"{self.nombre} ataca a {objetivo.nombre} por {dano_infligido} de daño.")
            objetivo.recibir_dano(dano_infligido)
```

```python
# --- Ejemplo de uso en un bucle de juego simple (simulado) ---

# Crear instancias de personajes (objetos)
jugador = Personaje("Heroe", 100, 0, 0, 5)
enemigo = Personaje("Ogro", 50, 10, 10, 2)
```

```python
# Simulación de acciones
jugador.mover(2, 2)  # El héroe se mueve
```

```python
jugador.atacar(enemigo) # El héroe ataca al ogro
```

```python
enemigo.atacar(jugador) # El ogro contraataca
```

```python
enemigo.recibir_dano(40) # El ogro recibe más daño y es derrotado
```

# Ejemplo 8 🍲 ⚕ ⚔ 🎮

```python
class Item:
    """Clase base para todos los objetos del juego."""
    def __init__(self, nombre, descripcion, peso):
        self.nombre = nombre
        self.descripcion = descripcion
        self.peso = peso

    def usar(self):
        """Método genérico para usar un ítem."""
        return f"Usaste el {self.nombre}."

# --- Clases que heredan de Item ---

class PocionSalud(Item):
    """Un ítem consumible que restaura salud."""
    def __init__(self, nombre, descripcion, peso, cantidad_curacion):
        # Llama al constructor de la clase padre (Item)
        super().__init__(nombre, descripcion, peso)
        self.cantidad_curacion = cantidad_curacion

    # Sobreescribe el método usar() de la clase padre
    def usar(self):
        return f"Bebiste la {self.nombre}. Curaste {self.cantidad_curacion} puntos de salud."

class Espada(Item):
    """Un arma que se puede equipar."""
    def __init__(self, nombre, descripcion, peso, dano_base):
        super().__init__(nombre, descripcion, peso)
        self.dano_base = dano_base

    # Sobreescribe el método usar() de la clase padre
    def usar(self):
        return f"Te equipaste la {self.nombre}. Daño de ataque aumentado en {self.dano_base}."

# --- Clase Gestor de Inventario ---

class Inventario:
    """Gestiona los ítems que lleva un personaje."""
    def __init__(self, capacidad_maxima_peso):
        self.capacidad_maxima = capacidad_maxima_peso
        self.items = [] # Una lista para guardar objetos Item
        self.peso_actual = 0

    def anadir_item(self, item_nuevo):
        """Intenta añadir un ítem al inventario."""
        if self.peso_actual + item_nuevo.peso <= self.capacidad_maxima:
            self.items.append(item_nuevo)
            self.peso_actual += item_nuevo.peso
            return f"Añadido {item_nuevo.nombre} al inventario."
        else:
            return f"No puedes añadir {item_nuevo.nombre}. El inventario está lleno."

    def listar_items(self):
        """Muestra el contenido actual del inventario."""
        print(f"--- Inventario ({self.peso_actual}/{self.capacidad_maxima} peso) ---")
        if not self.items:
            print("El inventario está vacío.")
        for item in self.items:
            print(f"- {item.nombre}: {item.descripcion}")


```

```python
# --- Simulación de uso ---

# 1. Crear objetos (ítems específicos)
pocion_pequena = PocionSalud("Poción pequeña", "Restaura un poco de vida", 0.1, 20)
espada_larga = Espada("Espada larga", "Filosa y pesada", 3.0, 15)
pocion_grande = PocionSalud("Poción grande", "Restaura mucha vida", 0.5, 50)

# 2. Crear un inventario para nuestro personaje
inventario_heroe = Inventario(capacidad_maxima_peso=3.5)
```

```python
# 3. Interactuar con el inventario
print(inventario_heroe.anadir_item(pocion_pequena))
```

# Ejemplo 9 🍠

 Aquí tienes un ejemplo de una clase Neurona que simula el comportamiento básico de una neurona en una red neuronal. Esta clase te ayudará a entender conceptos como pesos, sesgos y funciones de activación.

La clase Neurona recibe pesos y un sesgo en su inicialización. Su método predecir toma una lista de entradas, las multiplica por los pesos correspondientes, suma el sesgo y aplica una función de activación (en este caso, la función sigmoide) para producir una salida. Esto es una representación simplificada de cómo una neurona procesa la información.

1_v88ySSMr7JLaIBjwr4chTw.jpg

```python
import numpy as np

class Neurona:
    def __init__(self, pesos, sesgo):
        # 'pesos' son los valores que indican la importancia de cada entrada.
        # 'sesgo' es un valor adicional que ajusta la salida de la neurona.
        if not isinstance(pesos, (list, np.ndarray)) or not all(isinstance(p, (int, float)) for p in pesos):
            raise ValueError("Los pesos deben ser una lista o array de números.")
        if not isinstance(sesgo, (int, float)):
            raise ValueError("El sesgo debe ser un número.")

        self.pesos = np.array(pesos)
        self.sesgo = sesgo

    def sigmoide(self, x):
        # La función de activación Sigmoide convierte cualquier valor en un número entre 0 y 1.
        # Es útil para problemas de clasificación binaria.
        return 1 / (1 + np.exp(-x))

    def predecir(self, entradas):
        # 'entradas' es una lista de valores que la neurona recibe.
        if not isinstance(entradas, (list, np.ndarray)) or not all(isinstance(e, (int, float)) for e in entradas):
            raise ValueError("Las entradas deben ser una lista o array de números.")
        if len(entradas) != len(self.pesos):
            raise ValueError("El número de entradas debe coincidir con el número de pesos.")

        entradas_array = np.array(entradas)

        # 1. Calcular la suma ponderada: (entrada1 * peso1) + (entrada2 * peso2) + ... + sesgo
        suma_ponderada = np.dot(entradas_array, self.pesos) + self.sesgo

        # 2. Aplicar la función de activación (sigmoide en este caso)
        salida = self.sigmoide(suma_ponderada)
        return salida
```

```python
# --- Ejemplos de uso ---

# Definir los pesos y el sesgo para nuestra neurona
# Supongamos que tenemos dos entradas
pesos_ejemplo = [0.5, -0.2]
sesgo_ejemplo = 0.1

# Crear una instancia de la neurona
mi_neurona = Neurona(pesos_ejemplo, sesgo_ejemplo)
print(f"Neurona creada con pesos: {mi_neurona.pesos} y sesgo: {mi_neurona.sesgo}")
```

Predicciones: Se realizan tres predicciones con diferentes conjuntos de entradas ([0.8, 0.4], [0.1, 0.9], [0.9, 0.1]). Puedes observar cómo la salida de la neurona (un valor entre 0 y 1, gracias a la función sigmoide) cambia en función de las entradas y los pesos.

```python
# Probar con diferentes entradas
entradas_1 = [0.8, 0.4] # Primera entrada: 0.8, Segunda entrada: 0.4
salida_1 = mi_neurona.predecir(entradas_1)
print(f"Para entradas {entradas_1}, la salida de la neurona es: {salida_1:.4f}")
```

Finalmente, se demuestran los mecanismos de validación que implementamos en la clase:
